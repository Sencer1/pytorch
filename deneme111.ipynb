```
fdm = jsbsim.FGFDMExec(jsbsim.get_default_root_dir()) # simülasyon motoru
```

```
fdm.load_model("c172p")
```

```
fdm["ic/h-sl-ft"] = 5000 # deniz seviyesine göre yükskelik
fdm["ic/vc-kts"] = 120 # hava hızı
fdm["ic/terrain-elevation-ft"] = 0

fdm.run_ic() # verilen koşullara göre simülasyonu başlatıyor
```

```
altitude = fdm["position/h-sl-ft"]
velocity = fdm["velocities/vtrue-kts"]
pitch = fdm["attitude/pitch-rad"]
roll = fdm["attitude/phi-rad"] # sağ sol kanadın aşağı yukarı yatırması
yaw = fdm["attitude/psi-rad"] # uçağın burnunn sağa veya sola dömesi

print(altitude)
print(velocity)
print(pitch)
print(roll)
print(yaw)
```

```
for i in range(10):
  fdm.run()

  print("step:", i,
        "alt:", fdm["position/h-sl-ft"],
        "speed:", fdm["velocities/vtrue-kts"])
```

```
fdm["fcs/elevator-cmd-norm"] = 0.2

for i in range(20):
  fdm.run()

  print("step:", i,
        "pitch:", fdm["attitude/pitch-rad"], #burnun yukarı aşağı hareketi
        "altitude:", fdm["position/h-sl-ft"])

```

```
"""
elevator pitch i kontorl ediyor
aileron roll u kontorl ediyor
rudder yaw ı kontorl ediyor
throttle motor itkiyi kontorl ediyor
"""

fdm["fcs/elevator-cmd-norm"] = 0.0
fdm["fcs/aileron-cmd-norm"] = 0.3
```

```
for i in range(20):
  fdm.run()

  print("step:", i,
        "roll:", fdm["attitude/phi-rad"],
        "pitch:", fdm["attitude/theta-rad"])
```
